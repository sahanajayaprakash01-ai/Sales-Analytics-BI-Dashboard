```python
# 11. Sales by Year and Category
df.groupby(
    [df["Order Date"].dt.year, "Category"]
)["Sales"].sum()


# 12. Sales by Region and Category
df.groupby(["Region", "Category"])["Sales"].sum()


# 13. Sales by Segment and Region
df.groupby(["Segment", "Region"])["Sales"].sum()


# 14. Average Shipping Days by Ship Mode
df.groupby("Ship Mode")["Shipping Days"].mean().sort_values()


# 15. Maximum Shipping Days by Ship Mode
df.groupby("Ship Mode")["Shipping Days"].max()


# 16. Sales and Shipping Days by Ship Mode
df.groupby("Ship Mode").agg({
    "Sales": "sum",
    "Shipping Days": "mean"
})


# 17. Monthly Sales
df.groupby(df["Order Date"].dt.month)["Sales"].sum()


# 18. Yearly Sales
df.groupby(df["Order Date"].dt.year)["Sales"].sum()


# 19. Sales by Year and Month
df.groupby([
    df["Order Date"].dt.year,
    df["Order Date"].dt.month
])["Sales"].sum()


# 20. Correlation between Sales and Shipping Days
df["Sales"].corr(df["Shipping Days"])


# 21. Sales distribution statistics
df["Sales"].describe()


# 22. Top 10 highest sales records
df.nlargest(
    10, "Sales"
)[["Order ID", "Product Name", "Category", "Sales"]]


# 23. Lowest 10 sales records
df.nsmallest(
    10, "Sales"
)[["Order ID", "Product Name", "Category", "Sales"]]
```

```python
# 24. Sales by Category and Segment
df.groupby(["Category", "Segment"])["Sales"].sum().sort_values(ascending=False)


# 25. Top 10 States by Sales
df.groupby("State")["Sales"].sum().sort_values(ascending=False).head(10)


# 26. Top 10 Cities by Sales
df.groupby("City")["Sales"].sum().sort_values(ascending=False).head(10)


# 27. Top 10 Products by Sales
df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)


# 28. Sales by Year and Category
df.groupby(
    [df["Order Date"].dt.year, "Category"]
)["Sales"].sum()


# 29. Shipping Days statistics
df["Shipping Days"].describe()


# 30. Sales vs Shipping Days correlation
df["Sales"].corr(df["Shipping Days"])


# 31. Average Sales by Ship Mode
df.groupby("Ship Mode")["Sales"].mean().sort_values(ascending=False)


# 32. Average Shipping Days by Region
df.groupby("Region")["Shipping Days"].mean().sort_values(ascending=False)


# 33. Sales by Quarter
df.groupby(
    df["Order Date"].dt.quarter
)["Sales"].sum()


# 34. Number of orders by year
df.groupby(
    df["Order Date"].dt.year
)["Order ID"].nunique()


# 35. Number of unique customers
df["Customer ID"].nunique()
```

```python
#Sales by Category — Bar Chart
#This helps visually compare which category generates the most sales.
import matplotlib.pyplot as plt
import seaborn as sns

category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=category_sales.index, y=category_sales.values)
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
#Sales Trend by Year — Line Chart
year_sales = df.groupby(df["Order Date"].dt.year)["Sales"].sum()

plt.figure(figsize=(8,5))
plt.plot(year_sales.index, year_sales.values, marker="o")
plt.title("Sales Trend by Year")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```

```python
#Monthly Sales Trend
monthly_sales = df.groupby(
    df["Order Date"].dt.month
)["Sales"].sum()

plt.figure(figsize=(10,5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(range(1,13))
plt.grid(True)
plt.show()
```

```python
#Sales by Region
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=region_sales.index, y=region_sales.values)
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

```python
#Sales by Segment
segment_sales = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=segment_sales.index, y=segment_sales.values)
plt.title("Sales by Customer Segment")
plt.xlabel("Segment")
plt.ylabel("Total Sales")
plt.show()
```

```python
#Top 10 Sub-Categories
subcategory_sales = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=subcategory_sales.values, y=subcategory_sales.index)
plt.title("Top 10 Sub-Categories by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Sub-Category")
plt.show()
```

```python
#Sales vs Shipping Days
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Shipping Days", y="Sales")
plt.title("Sales vs Shipping Days")
plt.xlabel("Shipping Days")
plt.ylabel("Sales")
plt.show()
```
