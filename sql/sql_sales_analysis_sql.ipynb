```python
import pandas as pd
df=pd.read_csv("train.csv")
```

```python
import sqlite3

conn = sqlite3.connect("superstore.db")

df.to_sql("superstore", conn, if_exists="replace", index=False)

print("Table created successfully")
```

```python
pd.read_sql_query(
    "SELECT * FROM superstore LIMIT 5",
    conn
)
```

```python
pd.read_sql_query(
    """
    SELECT "Order ID", Sales
    FROM superstore
    LIMIT 10
    """,
    conn
)
```

```python
#Find all orders where Sales is greater than 1,000. Show the Order ID, Product Name and Sales.”
pd.read_sql_query(
    """
    SELECT "Order ID", Sales,"Product Name"
    FROM superstore
    where Sales >=1000
    """,
    conn
)
```

```python
#Good. Now I want to know which products generated the highest sales. Give me the top 10 records with the highest Sales. Show Product Name, Category, and Sales.”
pd.read_sql_query("""
select "Product Name", Category,Sales
from superstore
ORDER BY Sales DESC
LIMIT 10


""",conn
)
```

```python
#Can you give me the total sales for each product category? I want to see which category is generating the most revenue.”
pd.read_sql_query(
    """
    select sum(sales),Category
    from superstore
    group by Category
    """,conn
)
```

```python
#Good. Now sort those categories from highest sales to lowest sales. I want the highest-performing category at the top.
pd.read_sql_query(
    """
    select category,sum(Sales) as total_sales
    from superstore
    group by category
    order by total_sales ASC

    """,conn
)
```

```python
#We know which category performs best. Now I want to know which individual products are driving those sales.
#Give me the top 10 products based on total sales. Show the Product Name and total sales.
pd.read_sql_query("""
SELECT "Product Name", SUM(Sales) AS Total_sales
FROM superstore
GROUP BY "Product Name"
ORDER BY Total_sales DESC
LIMIT 10
""", conn)
```

```python
#Show me only the product categories whose total sales are greater than 700,000.
pd.read_sql_query(
    """
      select category ,sum(sales) as total_sales
      from superstore
      Group by category
      having sum(sales)>700000
    """,conn

)
```

```python
pd.read_sql_query("""
SELECT "Order ID", "Product Name", Category, Sales
FROM superstore
WHERE Category = 'Technology'
AND Sales > 1000
""", conn)
```

```python
#For each region, tell me how many unique customers we have.
pd.read_sql_query("""SELECT Region, COUNT(DISTINCT "Customer ID") AS Unique_Customers
FROM superstore
GROUP BY Region""", conn)
```

```python
pd.read_sql_query("""
SELECT Category, AVG(Sales) AS Average_Sales
FROM superstore
GROUP BY Category
ORDER BY Average_Sales DESC;
""", conn)
```

```python
#Classify every sales record as High, Medium, or Low based on Sales: above $1,000 = High, $100–$1,000 = Medium, below $100 = Low. Show the Sales amount and classification.”
pd.read_sql_query("""
SELECT
    Sales,
    CASE
        WHEN Sales > 1000 THEN 'High'
        WHEN Sales >= 100 THEN 'Medium'
        ELSE 'Low'
    END AS Sales_Level
FROM superstore;
""", conn)
```

```python
#Show me total sales for each year so we can understand our yearly performance
pd.read_sql_query("""
SELECT
    strftime('%Y', "Order Date") AS Year,
    SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Year
ORDER BY Year;
""", conn)
#SQLite uses strftime() for dates.
```

```python
#Find all products whose total sales are higher than the average product's total sales.”
pd.read_sql_query("""
SELECT
    "Product Name",
    SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY "Product Name"
HAVING SUM(Sales) > (
    SELECT AVG(Product_Total)
    FROM (
        SELECT SUM(Sales) AS Product_Total
        FROM superstore
        GROUP BY "Product Name"
    )
)
ORDER BY Total_Sales DESC;
""", conn)
```

```python
#“Within each category, rank the products based on their total sales. I want to know the best-performing products within each category.”
pd.read_sql_query("""
SELECT
    Category,
    "Product Name",
    SUM(Sales) AS Total_Sales,
    RANK() OVER (
        PARTITION BY Category
        ORDER BY SUM(Sales) DESC
    ) AS Sales_Rank
FROM superstore
GROUP BY Category, "Product Name"
ORDER BY Category, Sales_Rank;
""", conn)
```
